Compact branch-rollout smoke experiment for the public hackathon repository.

```python
import sys
from pathlib import Path
ROOT=Path.cwd()
if (ROOT/'src').exists(): sys.path.insert(0,str(ROOT/'src'))
import torch
print('torch',torch.__version__,'cuda',torch.cuda.is_available())
```

```python
from cidm_architectures import CIDMConfig
from parallel_future_engine import CIDMParallelFutureEngine, ParallelFutureConfig
from demo_physics import make_demo_history
device='cuda' if torch.cuda.is_available() else 'cpu'
cfg=CIDMConfig(channels=1,hist_len=4,horizon=8,latent_dim=16,hidden=32)
engine=CIDMParallelFutureEngine(cfg,ParallelFutureConfig(branches=5,horizon=8)).to(device).eval()
x=torch.from_numpy(make_demo_history(size=32)).to(device)
out=engine.branch_rollout(x)
print('pred',out['pred'].shape)
print('probabilities',out['branch_probability'].detach().cpu().numpy())
print('sum',float(out['branch_probability'].sum()))
```
